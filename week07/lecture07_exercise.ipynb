```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go

df = pd.read_csv('../data/netflix_catalogue.csv')
print(f"Loaded: {len(df)} titles")
print(df['type'].value_counts())
print(df.head())
```

```python
print("Genres:", df['genre'].value_counts().head(8))
print("\nCountries:", df['country'].value_counts().head(8))
print("\nRatings:", df['rating'].value_counts())
```

## Task 1 — Heatmap: content by rating and release decade

**What to build:** A heatmap showing the number of titles by **content rating** (y-axis) and **decade** (x-axis).

**Requirements:**
- Create a 'decade' column: `df['decade'] = (df['release_year'] // 10 * 10).astype(str) + 's'`
- Filter to TV-14, TV-MA, PG-13, R, PG only (most common ratings)
- Sequential colour scale (Blues)
- Values shown in cells (`text_auto=True`)
- Insight title about which rating dominates which decade


```python
# Task 1
# YOUR CODE HERE

df["decade"] = (df["release_year"] // 10 * 10).astype(str) + "s"

ratings_to_keep = ["TV-14", "TV-MA", "PG-13", "R", "PG"]

rating_decade = (
    df[df["rating"].isin(ratings_to_keep)]
    .groupby(["rating", "decade"])
    .size()
    .reset_index(name="count")
)

pivot = rating_decade.pivot(
    index="rating",
    columns="decade",
    values="count"
).fillna(0)

fig = px.imshow(
    pivot,
    text_auto=True,
    color_continuous_scale="Blues",
    labels={
        "x": "Release Decade",
        "y": "Content Rating",
        "color": "Number of Titles"
    },
    title="TV-MA and TV-14 dominate Netflix's newer catalogue"
)

fig.add_annotation(
    x="2010s",
    y="TV-MA",
    text="TV-MA peaks here",
    showarrow=True,
    arrowhead=2,
    ax=40,
    ay=-40
)

fig.update_layout(
    height=500,
    width=800
)

fig.show(renderer="png")
```

## Task 2 — Waterfall: Movie vs TV Show additions by year

**What to build:** A waterfall chart showing how Netflix's **Movie library** grew year by year (2015-2022).

**Requirements:**
- Filter to Movies only
- Group by `added_year`, count titles per year
- Final bar should be the cumulative total
- Green bars (additions), blue total
- Annotation on the year with the largest single addition
- Insight title naming the growth story


```python
# Task 2
# YOUR CODE HERE
movies = df[
    (df["type"] == "Movie") &
    (df["added_year"] >= 2015) &
    (df["added_year"] <= 2022)
].copy()

adds = (
    movies.groupby("added_year")
    .size()
    .reset_index(name="new_titles")
)

cumulative_total = adds["new_titles"].sum()

x_vals = adds["added_year"].astype(str).tolist() + ["Total"]
y_vals = adds["new_titles"].tolist() + [cumulative_total]
measure_vals = ["relative"] * len(adds) + ["total"]
text_vals = adds["new_titles"].astype(str).tolist() + [str(cumulative_total)]

max_year = adds.loc[adds["new_titles"].idxmax(), "added_year"]
max_value = adds["new_titles"].max()

fig = go.Figure(
    go.Waterfall(
        x=x_vals,
        y=y_vals,
        measure=measure_vals,
        text=text_vals,
        textposition="outside",
        connector=dict(line=dict(color="gray")),
        increasing=dict(marker=dict(color="green")),
        decreasing=dict(marker=dict(color="red")),
        totals=dict(marker=dict(color="blue"))
    )
)

fig.add_annotation(
    x=str(max_year),
    y=max_value,
    text=f"Largest addition: {max_value} movies",
    showarrow=True,
    arrowhead=2,
    ax=40,
    ay=-50
)

fig.update_layout(
    title="Netflix's movie catalogue grew fastest during its biggest annual additions",
    xaxis_title="Year Added",
    yaxis_title="Number of Movies Added",
    height=550,
    width=900,
    showlegend=False
)

fig.show(renderer="png")
```
